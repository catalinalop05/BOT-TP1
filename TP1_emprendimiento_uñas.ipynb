```python

turnos_disponibles=[1,2,3]
turnos_ocupados=[]


#diccionario
turno= {"dia":[],
        "servicio":[],
        "dni":[],
        "nombre":[] }

while len(turnos_disponibles) > 0:
 nombre= input("ingresa tu nombre")
 print(f"hola {nombre}, en que puedo ayudarte?")
 print( "1,reservar turno,2,cancelar turno,3,servicios y precios" )
 opciones = int(input("elige una de las opciones"))



 if opciones == 1:
  print("1,semi,2,capping,3,soft gel")
  servicio= input("¿que servicio queres agendar?")
  print(turnos_disponibles)
  dia= int(input("¿que dia/hora queres tu turno?"))
  dni= input("ingresa tu dni")


  turnos_ocupados.append(dia)
  turnos_disponibles.remove(dia)

  turno["dia"].append(dia)
  turno["dni"].append(dni)
  turno["servicio"].append(servicio)
  turno["nombre"].append(nombre)
  print(turno)
  print(f"listo,agendaste tu turno para el dia {dia}")

 elif opciones == 2:
   dni= input("ingresa tu dni")
   if dni in turno.values():
     print(f"tenes un turno agendado")
     cancelar= input("¿queres cancelar tu turno?")
     if cancelar.lower == "si":
       print("tu turno fue cancelado")
       turnos_ocupados.remove(dia)  #no se como hacer para ver el dia del turno
       turnos_disponibles.append(dia)

       turno["dia"].remove(dia)
       turno["dni"].remove(dni)
       turno["servicio"].remove(servicio)

     else:
       print("tu turno sigue agendado")
   else:
     print("no tenes turnos agendados")

 elif opciones == 3:
   print(f"1,semi $20,2,capping $30,3,soft gel $40")

 else:
   print("error, esa opcion no esta disponible, intentalo de nuevo")





```
